## Import Statements

```python
import pandas as pd
import seaborn as sb
```

**Sales:** Sales of a single product in a given market (in thousand of items)

There is a linear relationship between the features and response so it is a strong candidate for linear regression

## Linear Regression

### Pros

It is a popular modelling technique for **4** main reasons:

- It runs quickly (becomes increasingly important as the size of dataset increases)

- No tuning required (like we tuned KNN earlier)

- Highly interpretable and easy to understand and how it's making predictions

- Has been studied for many years, so it's well understood and a vast amount of documentation on how to use it properly

### Cons

- Unlikely to produce the best predictive accuracy as compared to other models because it presumes a linear relationship between the features and the response

- If the relationship is nonlinear like in alot of real world scenarios, in that case linear regression will be unable to effectively model the relationship hence it's predictions will not be very inaccurate

### Form of Linear Regression

y= B0 + B1x1 + B2x2 ... Bnxn

Here y is the response, B0 is the intercept, B1 is the co-efficient for x1 (first feature) and Bn is the coefficient for n feature

**Note:** B0(intercept) is the value of y when all the values of x are zero(0)

**Linear Regression VS Regression**

Regression is a type of supervised learning in which the response is continuos, while linear regression is a particular machine learning model that can be used for regression problems and it just happens to have regression in it's name

```python
# Preparing X and y using pandas
''' Remember, scikit-learn expects X (feature matrix) and y (response vector) to be numpy arrays but luckily
    pandas is built on top of numpy so X can be a pandas dataframe and y can be a pandas series '''
    
X= dataset[["TV","Radio","Newspaper"]]
X.head()

y= dataset[["Sales"]]
y.head()
```

## Splitting X and y into training and testing sets

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y, random_state=1)
```

```python
# Default split is 75% for training and 25% for testing
print(X_train.shape)
print(X_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
# Applying Linear Regression
from sklearn.linear_model import LinearRegression
lr=LinearRegression()
lr.fit(X_train,y_train)
```

```python
print (lr.intercept_)
print (lr.coef_)
lr.coef_.shape
```

```python
# Pair the features name with coef_
list((zip(["TV","Radio","Newspaper"],lr.coef_.ravel())))
```

## Now the equation is: 

### y= 4.64 + 0.0548 x TV + 0.1021 x Radio + 0.00078 x Newspaper

Interpreting the TV co-efficient (0.0548):

Increasing the TV by unit(one) would mathematically increase the y response variable by 0.0548

**Note:** Keep in mind that a unit of ad spending represents a 1000$ dollars and a unit of sales represents a 1000 items

A more clear interpretation of the TV co efficient would be that, an additional 1000$ spent on TV ads is associated with an increase in sales of 54.8 items. The same process would be repeated for the Radio and Newspaper co efficients

- This is a statement of association not causation because causation requires us to have access to every possible factor that could have influenced sales but all we have is data on ad spending, that's why machine learning works on association rather than causation

- Linear regression co efficients can be negative too for example: If an increase in TV ad spending was associated with a decrease in sales then the B1(Beta1) co efficient would have been negative

## Making Predictions

```python
y_pred= lr.predict(X_test)
```

Previously we have used classification accuracy as our evaluation metric but that metric is not relevant for regression problems because regression problems have a continuos response

## Model Evaluation metrics for Regression

Creating some example numeric predictions and calculating the three common evaluation metrics for regression problems



```python
real= [100, 50, 30, 20]
pred= [90, 50, 50, 30]
```

### 1- Mean Absolute Error

It is the mean of the absolute values of the errors (error is the difference between true and predicted value)

```python
# Calculating manually
print((10 + 0 + 20 + 10) / 4)

## Calculating using scikit-learn
from sklearn import metrics
print(metrics.mean_absolute_error(real,pred))
```

### 2- Mean Squared Error

It is the mean of the squared errors

```python
# Calculating manually
print((10**2 + 0**2 + 20**2 + 10**2) / 4)

## Calculating using scikit-learn
from sklearn import metrics
print(metrics.mean_squared_error(real,pred))
```

### 3- Root Mean Squared Error

It is similar to Mean squared error just a square root is taken in the end

```python
# Calculating manually
import numpy as np
print(np.sqrt((10**2 + 0**2 + 20**2 + 10**2) / 4))

## Calculating using scikit-learn
from sklearn import metrics
print(metrics.root_mean_squared_error(real,pred))
```

## Comparison

- **MAE** is the easiest to understand, because it's the average error
- **MSE** is more popular than MAE because MSE punishes large errors
- **RMSE** even more popular than MSE because RMSE is interpretable in the "y" units

We will choose **RMSE** Because MSE of 150 is a bit hard to interpret while the RMSE of around 12 is easier to put into content as it's in the same unit as our response variable 

## Computing the RMSE for our sales predictions

```python
print(np.sqrt(metrics.root_mean_squared_error(y_test,y_pred)))
```

Linear Regression does'nt have any tuning parameters and we haven't learned other models for regression yet, but note that train/test split can also help us to choose between features, when we visualized the data earlier we noted that the Newspaper feature had a very weak co-relation with sales so let's try removing it and how it affects our RMSE

### Removing Newspaper from the features and checking the RMSE

```python
X=dataset[["TV","Radio"]]
y=dataset[["Sales"]]

X_train,X_test,y_train,y_test=train_test_split(X,y,random_state=1)

lr.fit(X_train,y_train)

y_pred=lr.predict(X_test)

print(np.sqrt(metrics.root_mean_squared_error(y_test,y_pred)))
```

The RMSE slightly decreased when we removed "Newspaper". Now unlike classification accuracy where higher numbers are better, error is something that we want to minimize and thus lower numbers for RMSE are better thus the model which doesn't contain the feature "Newspaper" is performing slightly better than the model which contains it indicating that the "Newspaper" feature should be left out when training the model
