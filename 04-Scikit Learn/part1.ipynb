## Two types of Supervised Learning:
**1- Classification:**  A problem in which the outcome being predicted is categorical meaning its values are in a finite unordered set.

**2- Regression:**  A problem in which the outcome being predicted is ordered and continuos such as the price of a house or height of a person.

Predicting the species of iris is a Classification problem

## 4 Requirements of working with data in Scikit-Learn

1- Features(column) and response(target) are seperate objects

2- Both of them should be numeric

3- Both of them should be numpy arrays

4- Both of them should have specific shapes


```python
iris.data.shape

# Output: (No: of rows, no: of columns)

iris.target.shape

# Output: (No: of rows,)
```

```python
X = iris.data # Capitalized because it represents a matrix
y = iris.target # Lowercase because it represents a vector
```

## Scikit-Learn 4 step Modelling pattern

Using K- nearest neighbors (KNN) Classification Model for this example

- **Step 1**: Import the class you plan to use
- **Step 2**: Instantiate(make an instance of) the estimators(models)
- **Step 3**: Fit the model with data (model training)
- **Step 4**: Predict the response for a new observation

```python
from sklearn.neighbors import KNeighborsClassifier
knn= KNeighborsClassifier(n_neighbors=1)
# n_neighbors tells the object when it runs it should be looking for one nearest neighbor
knn # shows all the hyperparameters that are set by default and are customizable
```

```python
knn.fit(X,y)
```

```python
knn.predict([[3,5,4,2]])
```

Two new observations:

```python
X_new=[[3,5,4,2],[5,4,3,2]] 
knn.predict(X_new)
```

## Using a different value for k (n_neighbors)

```python
knn_2=KNeighborsClassifier(n_neighbors=5)
knn_2.fit(X,y)
knn_2.predict(X_new)
```

## Trying a different classification model on this problem
(the 4 steps remain same)

```python
from sklearn.linear_model import LogisticRegression

lr=LogisticRegression()
lr.fit(X,y)
lr.predict(X_new)
```
