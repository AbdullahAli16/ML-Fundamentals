# How to decide which model to use and when ?

## Import Statments

```python
from sklearn.linear_model import LogisticRegression
from sklearn.datasets import load_iris
```

How would we choose the best tuning parameters for that model ?

How would we evaluate the current performance of our model on the desired data?

We will do that by evaluating the performace of the models.

**Evaluation Procedure 1:** Train and test on the entire dataset

We will test the model on the same dataset and compare its predicted response(target) values with the true response(target) values

```python
iris= load_iris()
X= iris.data
y= iris.target
lr= LogisticRegression()
lr.fit(X,y)
y_pred=lr.predict(X)
len(y_pred)
```

```python
from sklearn import metrics
metrics.accuracy_score(y,y_pred)
# Output: 0.973~ meaning approximately 97% of our predictions are correct
```

Training and testing the model on the same data ia known as **training_accuracy** 

Maximizing **Training accuracy** rewards overly complex models that are unlikely to generalize duture data which is known as "overfitting", because remember the goal of supervised learning is to **build a model that generalizes to out of sample data**

Using KN neighbors Model on the data

KNN(1)

```python
from sklearn.neighbors import KNeighborsClassifier
knn= KNeighborsClassifier(n_neighbors=1)
knn.fit(X,y)
y_pred=knn.predict(X)
metrics.accuracy_score(y,y_pred)
```

KNN(5)

```python
knn=KNeighborsClassifier(n_neighbors=5)
knn.fit(X,y)
y_pred=knn.predict(X)
metrics.accuracy_score(y,y_pred)
```

**Evaluation Procedure: 2** Train, test and split

1. Split the dataset into 2 pieces: training set and testing set
2. Train the model using the training set
3. Test the model using testing set, and evaluate how well it performed
( The resulting evaluation metric is known as **testing accuracy** and it is a better estimate then training accuracy of out-of-sample performance )

```python
print(X.shape)
print(y.shape)
```

```python
# Step 1
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=0.4, random_state=4)
```

```python
print(X_train.shape)
print(X_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
# Step 2
from sklearn.linear_model import LogisticRegression
lr=LogisticRegression()
lr.fit(X_train,y_train)
```

```python
# Step 3
y_pred=lr.predict(X_test)
metrics.accuracy_score(y_test,y_pred)
```

Doing the same with KNN(1) and KNN(5)

```python
knn=KNeighborsClassifier(n_neighbors=1)
knn.fit(X_train,y_train)
y_pred=knn.predict(X_test)
metrics.accuracy_score(y_test,y_pred)
```

```python
knn=KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train,y_train)
y_pred=knn.predict(X_test)
metrics.accuracy_score(y_test,y_pred)
```

Trying to find an even better k value:

```python
k_range=range(1,26)

accuracy_scores=[]
for k in k_range:
    knn=KNeighborsClassifier(n_neighbors=k)
    knn.fit(X_train,y_train)
    y_pred=knn.predict(X_test)
    accuracy_scores.append(metrics.accuracy_score(y_test,y_pred))
accuracy_scores
```

Visualizing it using matplotlib

```python
from matplotlib import pyplot as plt

plt.plot(k_range,accuracy_scores)
plt.xlabel("Values of K for KNN")
plt.ylabel("Testing Accuracy")
```

Picking k=11 value as it is providing the best testing accuracy and we will call that the best model so now we are training it on the entire data instead of just a fraction of it and making a prediction on out of sample data 

```python
knn=KNeighborsClassifier()
knn.fit(X,y)
knn.predict([[3, 5, 4 ,2]])
```

A downside of train/test split is that it provides a high variance(difference) of out of sample accuracy meaning it varies alot depending on the observations happen to be in the training versus the testing set, there is an alternative procedure called "K-fold cross-validation" which overcomes this limitation, but train/test split is still useful because of its flexibility and speed
